```python
import pandas as pd
print(pd.__version__)
```

```python
df = pd.read_csv("car_fuel_efficiency_2026.csv")

print(f"Number of records: {len(df)}")
```

```python
distinct_fuel_types = df["fuel_type"].nunique()

print(f"Number of distinct fuel types: {distinct_fuel_types}")
```

```python
missing_columns = df.isnull().any().sum()

print(f"Number of columns with missing values: {missing_columns}")
```

```python
max_fuel_efficiency = df["fuel_efficiency_mpg"].max()

print(max_fuel_efficiency)
```

```python
# 1. Find the original median
original_median = df["horsepower"].median()
print("Original median:", original_median)

# 2. Find the most frequent (mode) value
most_frequent = df["horsepower"].mode()[0]
print("Most frequent value:", most_frequent)

# 3. Fill missing values with the most frequent value
df["horsepower"] = df["horsepower"].fillna(most_frequent)

# 4. Calculate the median again
new_median = df["horsepower"].median()
print("New median:", new_median)

# 5. Check whether the median increased, decreased, or stayed the same
if new_median > original_median:
    print("The median increased.")
elif new_median < original_median:
    print("The median decreased.")
else:
    print("The median remained the same.")
```

```python
import numpy as np

# 1. Select cars from Asia
asia = df[df["origin"] == "Asia"]

# 2. Select vehicle_weight and model_year
X = asia[["vehicle_weight", "model_year"]]

# 3. Select first 7 values
X = X.head(7)

# 4. Get the underlying NumPy array
X = X.to_numpy()

print("X:")
print(X)

# 5. Calculate X^T X
XTX = X.T @ X

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Calculate w
w = XTX_inv @ X.T @ y

print("w:")
print(w)

# 9. Sum all elements of w
result = w.sum()

print("Sum of w:", result)
```
